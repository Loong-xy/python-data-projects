**Data**
- $I$: set of Lakers players (Integer)
- $P$: set of plays (Integer)
- $r$ set of offensive roles (Integer)
- $k$: set of defensive roles (Integer)
- $m_{pr}$: minimum requirements for role $r$ in play $p$ (Integer)

- $O_{i,p}$: expected points per possession (PPP) for i in play p (Continuous)
- $D_{i,k}$: expected points per possession (PPP) for i in defensive role k (Continuous)
- $C_{i1,i2}$: chemistry bonus for players i1 and i2 on court (Continuous)
- $V_{i,r}$: value of i in offensive role r (Continuous)

**Decision Variables**
- $X_i$: whether to play player i (Binary)
- $Y_{i1, i2}$: whether players i1 and i2 are playing together (Binary)
- $A_{i,r}$: whether player i is playing offensive role r (Binary)
- $Z_{i,k}$: whether player i is playing defensive role k (Binary)

**Objective:** Maximize the point differential in a lineup against matchups
$$\begin{aligned}
\text{Maximize:} \quad && \sum_{i,p \in F}\ O_{i,p}\, x_{i} - \sum_{i,k \in F}\ D_{i,k}\, z_{i,k} + \sum_{i,j \in F}\ C_{i,j}\, Y_{i1,i2} + \sum_{i,r \in F}\ V_{i,r}\, A_{i,r}
\end{aligned}$$

**Constraints:**
$$\begin{aligned}
\text{Five Players} && \sum_{i \in I} x_i = 5 \\
\text{Pre-req for Chemistry} && \ Y_{i_1,i_2} \le X_{i1}  \\
\text{Pre-req for Chemistry} && \ Y_{i_1,i_2} \le X_{i2}  \\
\text{Each player needs a defensive role} && \ \sum_{k \in K} Z_{i,k} = X_{i} && \text{for each $i \in I$.}\\
\text{Each defensive role can only be assigned once} && \ \sum_{k \in K} Z_{i,k} = 1 && \text{for each $k \in K$.}\\
\text{Optionally assign offensive roles to players on court} && \ \sum_{r \in R} a_{i,r} \le X_{i} && \text{for each $i \in I$.}\\
\text{Offensive role minimums are met for the selected play} && \ \sum_{r \in R} a_{i,r} \ge m_{pr} && \text{for each $i \in I$.}\\
\end{aligned}$$
